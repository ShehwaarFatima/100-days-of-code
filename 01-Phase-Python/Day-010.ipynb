Create a tuple named student_info containing a name, roll number, and 3 marks. Access and print the first mark and the last mark using positive and negative indexing.

```
student_info = ("Sanabil", 101, 85, 90, 88)

# Accessing first mark (index 2) and last mark (index -1)
first_mark = student_info[2]
last_mark = student_info[-1]

print("First Mark:", first_mark)
print("Last Mark:", last_mark)
```

Write a Python program to pack 4 values into a tuple, then unpack them into individual variables and print each variable.

```
# Tuple Packing
data = ("Python", 3.11, 2026, True)

# Tuple Unpacking
lang, version, year, status = data

print("Language:", lang)
print("Version:", version)
print("Year:", year)
print("Active:", status)
```

Given a tuple numbers = (10, 20, 30, 20, 40, 20, 50), write code to find:

How many times 20 appears in the tuple.

The index of the first occurrence of 40.

Python

```
numbers = (10, 20, 30, 20, 40, 20, 50)

# 1. Count occurrences
count_20 = numbers.count(20)

# 2. Find index
index_40 = numbers.index(40)

print("Count of 20:", count_20)
print("Index of 40:", index_40)
```

Extract and print the subject name "Anesthesia" and the score 95 from the following nested tuple:

```
records = ("Semester 4", ("Anesthesia", 95), ("Pathology", 88))

# Accessing nested elements
subject = records[1][0]
score = records[1][1]

print("Subject:", subject)
print("Score:", score)
```
