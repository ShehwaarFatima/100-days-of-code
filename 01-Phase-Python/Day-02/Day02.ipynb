```python
# if Statement
num = 10
if num > 5:
    print("Number is greater than 5")
```

```python
# if-else Statement
age = 16
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
# if-elif-else Statement
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Grade: C or below")
```

```python
# Nested if Statement
num = 12
if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")
```

```python
# Logical Operators
age = 20
has_id = True

if age >= 18 and has_id:
    print("Access Granted")
```

```python
# Comparison Operators in Conditions
age = 20
marks = 85

# Equal to (==)
if age == 20:
    print("Age is exactly 20.")

# Not equal to (!=)
if age != 18:
    print("Age is not 18.")

# Greater than (>) and Less than (<)
if marks > 80:
    print("Distinctive marks!")

# Greater than or equal to (>=) / Less than or equal to (<=)
if age >= 18:
    print("Eligible for driving license.")
```

```python
# Logical Operators in Conditions
score = 88
attendance = 92

# Both conditions must be True
if score >= 80 and attendance >= 90:
    print("Eligible for Merit Scholarship!")
```

```python
# The or Operator
day = "Sunday"
is_holiday = False

# Executes if either day is Sunday OR is_holiday is True
if day == "Sunday" or is_holiday:
    print("No class today, enjoy your day off!")
```

```python
#The not Operator
is_logged_in = False

# Evaluates to True because not False == True
if not is_logged_in:
    print("Please log in to access your account.")
```

```python
# Membership Operators in Conditions
allowed_users = ["admin", "instructor", "student"]
current_user = "instructor"

# Checking presence
if current_user in allowed_users:
    print("Access Granted.")

# Checking absence
banned_words = ["spam", "hack", "bot"]
comment = "Hello, this is a great project!"

if "spam" not in comment:
    print("Comment posted successfully.")
```

```python
# Identity Operators in Conditions
user_input = None

if user_input is None:
    print("No input was provided.")

value = 100
if value is not None:
    print(f"Processing value: {value}")
```

```python
# Arithmetic Operators inside Conditions
num = 14

# Using Modulus (%) to check if a number is Even
if num % 2 == 0:
    print(f"{num} is an Even number.")

# Checking discounted price calculation
original_price = 120
if (original_price - 20) < 100:
    print("Discounted price is under $100!")
```

```python
# Real-World Combined Example (Grade & Role Eligibility)
role = "student"
marks = 78
has_disciplinary_action = False

# Combining Logical, Comparison, and Identity operators
if role == "student" and (marks >= 70 and not has_disciplinary_action):
    print("Passed with Distinction and eligible for Honor Roll!")
elif role == "student" and marks >= 50:
    print("Passed standard evaluation.")
else:
    print("Requirements not met.")
```
