```
tup = ()
print(tup)

# Using String
tup = ('hello', 'word')
print(tup)

# Using List
li = [1, 2, 4, 5, 6]
print(tuple(li))

# Using Built-in Function
tup = tuple('hello')
print(tup)
```

```
tup = (5, 'Welcome', 7.5, True, [1, 2, 3], {'key': 'value'})
print(tup)
```

```
tup = tuple("hi")
print(tup[0])
print(tup[1:4])
print(tup[:3])

# Tuple unpacking
tup = ("hi", "hello", "hi")

# This line unpack values of Tuple1
a, b, c = tup
print(a)
print(b)
print(c)
```

```
tup = tuple('HowAreYou')
print(tup[1:])
print(tup[::-1])
print(tup[4:9])
```

```
tup = (0, 1, 2, 3, 4)
del tup
print(tup)
```

```
tup = (1, 2, 3, 4, 5)
a, *b, c = tup
print(a)
print(b)
print(c)
```
