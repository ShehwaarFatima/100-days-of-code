```
#Creating a String
a = 'GFG'
b = "GeeksForGeeks"
print(a)
print(b)
```

```
##Multi-line Strings
s = """I am Learning
Python String on GeeksforGeeks"""
print(s)

s = '''I'm a
Geek'''
print(s)
```

```
#Accessing Characters in String
s = "ABCDEF"
print(s[0])
print(s[4])
```

```
#reading Characters in String
s = "ABCDEF"
print(s[-3])
print(s[-5])
```

```
#String Slicing
s = "ABCDEF"
print(s[1:4])
print(s[:3])
print(s[3:])
print(s[::-1])
```

```
#Looping Through Strings
s = "ABCDEF"
for char in s:
    print(char)
```

```
#String Immutability
s = "aBCDEF"
s = "A" + s[1:]
print(s)
```
